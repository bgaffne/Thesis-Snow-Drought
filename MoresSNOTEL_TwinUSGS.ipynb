```python
from datetime import datetime
import pandas as pd

from metloom.pointdata import SnotelPointData, CDECPointData, USGSPointData
```

```python
# Let's look at our old friend banner summit
pt = SnotelPointData("637:ID:SNTL", "Mores Creek Summit")

# start data and end date
start_date = datetime(1986, 3, 1)
end_date = datetime(2026, 4, 1)
# Notice this is a list
variables = variables = [
    pt.ALLOWED_VARIABLES.SWE,
    pt.ALLOWED_VARIABLES.PRECIPITATIONACCUM,
    pt.ALLOWED_VARIABLES.RH,
    pt.ALLOWED_VARIABLES.TEMP,
]

# request the data
df = pt.get_daily_data(start_date, end_date, variables)
df.head(10)
```

```python
df.reset_index().set_index("datetime")["SWE"].plot()
```
